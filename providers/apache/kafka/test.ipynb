```python
# Define a simple test function for the trigger to use
def noop(message):
    """Simple test function that does nothing but return arguments."""
    return message


# Create a module-like object to make the function accessible
import sys
import types

from airflow.providers.apache.kafka.triggers.await_message import AwaitMessageTrigger
from airflow.providers.common.messaging.triggers.msg_queue import MessageQueueTrigger

test_module = types.ModuleType("test")
test_module.noop = noop
sys.modules["test"] = test_module

# Add Kafka connection

# Create Kafka trigger
trigger = MessageQueueTrigger(
    queue="kafka://localhost:9092/test_topic",
    apply_function ="test.noop",
)

print("Kafka trigger created successfully!")
```

```python
classpath, kwargs = trigger.trigger.serialize()
print(classpath)
print(kwargs)
```
